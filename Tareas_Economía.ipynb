##Calculadora.

```python
def calculadora(a, b, operacion):
  try:
    if operacion == '+':
      op = a + b
    elif operacion == '-':
      op = a - b
    elif operacion == '*':
      op = a * b
    elif operacion == '/':
      # Added a check for division by zero
      if b == 0:
        print('Error: Division por cero')
        return None # Return None or handle the error as appropriate
      op = a / b
    elif operacion == '%':
      op = a % b
    elif operacion == '**':
      op = a ** b
    else:
      print('Operacion no valida')
      return None  # Return None when the operation is invalid
    return op
  except ValueError:
    # This except block will not be reached by the current code but is kept
    # in case the inputs 'a' or 'b' were user input and not yet converted.
    print('El valor ingresado no es un numero')
    return None # Return None in case of a ValueError
  except ZeroDivisionError:
    # This except block is now redundant due to the explicit check for b == 0,
    # but it can be kept as a safeguard or removed.
    print('El valor ingresado no es un numero')
    return None # Return None in case of a ZeroDivisionError
```

##Conversor de temperatura.

```python

def convertir_temperatura(valor, unidad_origen, unidad_destino):
    try:
        valor = float(valor)
        unidad_origen = unidad_origen.upper()
        unidad_destino = unidad_destino.upper()

        if unidad_origen not in ['C', 'F', 'K'] or unidad_destino not in ['C', 'F', 'K']:
            raise ValueError("Unidad inválida. Usa 'C', 'F' o 'K'.")

        if unidad_origen == 'K' and valor < 0:
            raise ValueError("La temperatura en Kelvin no puede ser negativa.")
        if unidad_origen == 'C' and valor < -273.15:
            raise ValueError("La temperatura en Celsius no puede ser menor que -273.15.")
        if unidad_origen == 'F' and valor < -459.67:
            raise ValueError("La temperatura en Fahrenheit no puede ser menor que -459.67.")

        # Conversión a Celsius
        if unidad_origen == 'F':
            celsius = (valor - 32) * 5/9
        elif unidad_origen == 'K':
            celsius = valor - 273.15
        else:
            celsius = valor

        # Conversión desde Celsius
        if unidad_destino == 'F':
            return celsius * 9/5 + 32
        elif unidad_destino == 'K':
            return celsius + 273.15
        else:
            return celsius
    except ValueError as e:
        return f"Error: {e}"
    except Exception:
        return "Error inesperado."
```

##Generador contrasenas.

```python
import random
import string

def generar_contraseña(longitud):
    try:
        longitud = int(longitud)
        if longitud < 6:
            raise ValueError("La longitud mínima es 6.")

        caracteres = string.ascii_letters + string.digits + string.punctuation
        contraseña = ''.join(random.choice(caracteres) for _ in range(longitud))
        return contraseña
    except ValueError as e:
        return f"Error: {e}"
    except Exception:
        return "Error inesperado al generar la contraseña."


```

##Estadisticas.

```python
import statistics

def estadisticas(lista):
    try:
        if not lista:
            raise ValueError("La lista está vacía.")
        lista_numerica = [float(x) for x in lista]
        promedio = sum(lista_numerica) / len(lista_numerica)
        mayor = max(lista_numerica)
        menor = min(lista_numerica)
        mediana = statistics.median(lista_numerica)
        return {
            'promedio': promedio,
            'mayor': mayor,
            'menor': menor,
            'mediana': mediana
        }
    except ValueError as e:
        return f"Error: {e}"
    except Exception:
        return "Error: Asegúrate de que todos los elementos sean numéricos."
```

##Usuario y contra.

```python

import re

def validar_usuario_contraseña(usuario, contraseña):
    try:
        if not usuario or not contraseña:
            raise ValueError("Usuario o contraseña vacíos.")

        if len(usuario) < 5:
            raise ValueError("El usuario debe tener al menos 5 caracteres.")

        if len(contraseña) < 8:
            raise ValueError("La contraseña debe tener al menos 8 caracteres.")
        if not re.search(r"[A-Z]", contraseña):
            raise ValueError("La contraseña debe incluir al menos una mayúscula.")
        if not re.search(r"[0-9]", contraseña):
            raise ValueError("La contraseña debe incluir al menos un número.")

        return "Usuario y contraseña válidos."
    except ValueError as e:
        return f"Error: {e}"
    except Exception:
        return
```

```python
import math

def area_y_volumen_esfera():
    try:
        r = float(input("Introduce el radio de la esfera: "))
        if r < 0:
            raise ValueError("El radio no puede ser negativo.")
        area = 4 * math.pi * math.pow(r, 2)
        volumen = (4 / 3) * math.pi * math.pow(r, 3)
        print(f"Área: {area:.3f}")
        print(f"Volumen: {volumen:.3f}")
    except ValueError as e:
        print("Error:", e)

area_y_volumen_esfera()
```

```python
from datetime import datetime, timedelta

def calcular_vencimiento():
    try:
        fecha_str = input("Ingresa la fecha de inicio (YYYY-MM-DD): ")
        dias = int(input("Duración del plazo (días): "))
        fecha_inicio = datetime.strptime(fecha_str, "%Y-%m-%d")
        vencimiento = fecha_inicio + timedelta(days=dias)
        print("Fecha de vencimiento:", vencimiento.strftime("%d de %B de %Y"))
    except Exception as e:
        print("Error:", e)

calcular_vencimiento()
```

##Ejercicio 9.

```python
import math

def ahorro_mensual():
    try:
        aporte = float(input("Aporte mensual: "))
        meses = int(input("Número de meses: "))
        tasa_mensual = 0.01  # 1% mensual
        total = 0
        for i in range(1, meses + 1):
            total += aporte * math.pow(1 + tasa_mensual, meses - i)
        print(f"Monto total ahorrado: ${total:.2f}")
    except Exception as e:
        print("Error:", e)

ahorro_mensual()
```
